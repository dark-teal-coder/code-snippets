However, this augments, but does not replace, my existing version.  

The final output from running this command was this:

```python

#
#     $ conda activate py39
#
# To deactivate an active environment, use
#
#     $ conda deactivate
```

## Environments

This allows me to use two different environments

I can activate the py39 enviroment and use Python 3.9.

I can deactivate the environment and use Python 3.8.

It is easy to change at the command line

To control the version in a Notebook, you run the right environment when you start the notebook server.

To run this notebook in py39, I changed (cd) to the directory with the notebook and typed the following at the command line

```python
% conda activate py39
% jupyter notebook
```

## How does the magic work?

The activate and deactive commands manipulate the PATH variable.

````python
! which python
````

## Before activation

```python
% echo $PATH
/Users/jparker/opt/anaconda3/bin:/Users/jparker/opt/anaconda3/condabin:/usr/local/bin:/usr/bin:/bin:/usr/sbin:/sbin:/Library/TeX/texbin:/opt/X11/bin:/usr/local/git/bin:/Users/jparker/bin

% which python
/Users/jparker/opt/anaconda3/bin/python

% python
Python 3.8.10 (default, May 19 2021, 11:01:55) 
[Clang 10.0.0 ] :: Anaconda, Inc. on darwin
Type "help", "copyright", "credits" or "license" for more information.
>>> quit()
```

Notice that the python we are using was found in the first directory listed in the PATH

## After Activation

```python
% conda activate py39

% echo $PATH
/Users/jparker/opt/anaconda3/envs/py39/bin:/Users/jparker/opt/anaconda3/condabin:/usr/local/bin:/usr/bin:/bin:/usr/sbin:/sbin:/Library/TeX/texbin:/opt/X11/bin:/usr/local/git/bin:/Users/jparker/bin

% which python
/Users/jparker/opt/anaconda3/envs/py39/bin/python

 % python
Python 3.9.4 (default, Apr  9 2021, 09:32:38) 
[Clang 10.0.0 ] :: Anaconda, Inc. on darwin
Type "help", "copyright", "credits" or "license" for more information.
>>> 
```

The activate command changed the initial path to point to ... envs/py39/bin - a directory holding the 3.9 version of Python

## The importance of the PATH

The activate and deactivate commands just manipulate the PATH, directing you to different version of Python

I have a dozen versions of Python on my machine, but just use the most recent two.
